### Zaawansowane funkcje z użyciem `plotly.graph_objects`

Figury są tworzone w plotly za pomocą strukury danych `graph_objects`. Plotly express, pozwlala na szybkie tworzenie wykresów, jednak nie mamy wszystkich możliwości ich edycji. Niestety, nie działa to już tak fajnie z pandasem.

```python
import plotly.graph_objects as go
```

#### Zmiana template'u tekstu pojawiającego się po najechaniu

```python
fig = go.Figure()

obj1 = go.Scatter(x = df.Attack,
                  y = df.Defense,
                  mode = "markers",
                  marker = {"color": df["HP"]},
                  
                  text = [f"Ta kropka dotyczy pokemona {df.loc[i, 'Name']} o ataku {df.loc[i, 'Attack']}" for i in df.index],
                  
                  hovertemplate = """<i>%{text}</i><br>
                  <b>%{x:.2f}</b> 
                  <extra>A jego obrona to %{y:.1f}</extra>
                  """)

fig.add_trace(obj1)
fig.update_layout(title = "Attack vs. Defense vs. HP")
fig.update_xaxes(title = "Attack")
fig.update_yaxes(title = "Defense")

fig.show()

```

#### Wyłączenie guzików

```python
fig = go.Figure()

obj1 = go.Scatter(x = df.Attack,
                  y = df.Defense,
                  mode = "markers",
                  marker = {"color":df["HP"]})

fig.add_trace(obj1)
fig.update_layout(title = "Attack vs. Defense vs. HP")
fig.update_xaxes(title = "Attack")
fig.update_yaxes(title = "Defense")

fig.show(config = {"modeBarButtonsToRemove": ["toImage", "hoverClosestPie", "zoom2d", "pan2d", "select2d", "lasso2d", "zoomIn2d", "zoomOut2d", "autoScale2d", "resetScale2d"]})

```

```python
X = credit.data # informacje o kredytobiorcach
y = credit.target # infomracje o tym czy klient jest "dobry"/"zły"
```

```python
df = pd.concat([X, y], axis = 1)
```

## Hipotezy (Eksploracyjna Analiza Danych)
----

### Wiek a ryzyko kredytowe
------
*Hipoteza*: Młodsi klienci częściej mają problem ze spłatą kredytu niż klienci starsi (wraz z wiekiem rośnie stabilność życiowa).

### Status mieszkaniowy a ryzyko kredytowe
-----
*Hipoteza*: Osoby wynajmujące mieszkanie (rent) lub mieszkające za darmo (for free) generują wyższe ryzyko kredytowe niż właściciele nieruchomości (own).

### Kwota kredytu a długość jego trwania
-----
*Hipoteza*: Im wyższa kwota kredytu (credit_amount), tym dłuższy okres spłaty (duration), a najwięcej „złych” klientów koncentruje się wokół długoterminowych kredytów na wysokie kwoty.

### Cel kredytu a historia kredytowa
----
*Hipoteza*: Kredyty konsumpcyjne na bieżące wydatki (np. edukacja, wakacje, mały sprzęt AGD) są częściej brane przez osoby z gorszą historią kredytową niż kredyty inwestycyjne (np. samochód, biznes).

### Staż pracy a stan konta oszczędnościowego
-----
*Hipoteza*: Osoby o dłuższym stażu u obecnego pracodawcy (employment) posiadają większe oszczędności (savings_status), co bezpośrednio przekłada się na lepszy status kredytowy.
